```python
from linearframework.linear_framework_graph import LinearFrameworkGraph
from linearframework.gen_graphs import gen_random_edge_to_weight
from linearframework.linear_framework_results import steady_states
```

This note is a tutorial on the creation and format of LinearFrameworkGraph objects in linearframework.py.

In the Linear Framework, we use weighted, directed graphs as representations of continuous-time Markov processes.
Each vertex represents a particular markov state, edges represent transitions between states, and edge weights represent transition rates.

LinearFrameworkGraph objects initialized from a dictionary where edges point to weights. The datatypes of vertex id's are entirely arbitrary.

```python
k3_edge_to_weight = {
    ('1', '2'): 1,
    ('1', '3'): 2,
    ('2', '1'): 3,
    ('2', '3'): 4,
    ('3', '1'): 5,
    ('3', '2'): 6,
}

k3 = LinearFrameworkGraph(k3_edge_to_weight)
```

It may seem a little strange to initialize directed weighted graphs with no weights. This is because all of our methods are implemented symbolically--so each edge is assigned a symbol, which we can then modify later to evaluate at a point or perform other tasks.

LinearFrameworkGraph objects have the following attributes:
- `self.edges` (list of edges, where edges are tuples with two objects, each object is the name of a vertex in the graph)
- `self.edge_to_weight` (dictionary with edges as keys, pointing to edge label)
- `self.nodes` (list of vertices)
- `self.terminal_nodes` (list of terminal vertices)
- `self.terminal_edges` (list of edges leading to terminal vertices)

You can also generate a random `edge_to_weight` given a list of edges using the `gen_random_edge_to_weight` function. Try running the cell multiple times and see that the edge weights change.

```python
k3_edges = [
    ('1', '2'),
    ('1', '3'),
    ('2', '1'),
    ('2', '3'),
    ('3', '1'),
    ('3', '2'),
]

k3_random_edge_to_weight = gen_random_edge_to_weight(k3_edges)

k3_random = LinearFrameworkGraph(k3_random_edge_to_weight)

k3_random.edge_to_weight
```

You can also place in a random seed if you want to be able to perfectly reproduce your results. Try running the cell multiple times and see that the generated numbers don't change.

```python
k3_random_seed_edge_to_weight = gen_random_edge_to_weight(k3_edges, random_seed=1)

k3_random_seed = LinearFrameworkGraph(k3_random_seed_edge_to_weight)

k3_random_seed.edge_to_weight
```

Of course, this graph has no terminal vertices. Let's define a similar graph which does.

```python
terminal_graph_edges = [
        ('1', '2'),
        ('1', '3'),
        ('2', '1'),
        ('2', '3'),
        ('3', '1'),
        ('3', '2'),
        ('2', '4'),
        ('3', '5')
    ]

terminal_graph_edge_to_weight = gen_random_edge_to_weight(terminal_graph_edges)

terminal_graph = LinearFrameworkGraph(terminal_graph_edge_to_weight)

print(f'terminal nodes: {terminal_graph.terminal_nodes}')
print(f'terminal edges: {terminal_graph.terminal_edges}')
```

With graphs defined, we can do many things. For example, calculate the steady states. Most other results are included in other tutorials. This calculation makes use of the Matrix-Tree theorem.
